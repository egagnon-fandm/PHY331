# Directional derivatives

## 2D scalar field

```python
import numpy as np 
import matplotlib.pyplot as plt 

dx = 0.1
dy = 0.1
X, Y = np.meshgrid(np.arange(-5,5,dx), np.arange(-5, 5, dy))

def T(X, Y):
    return 10*np.exp(-(np.pow(X+1,2)+np.pow(Y-1,2))/2) + 4*np.exp(-(np.pow(X-1,2)+np.pow(Y,2))/3)

fig = plt.figure()
ax = fig.add_subplot(111)
ax.imshow(T(X,Y), origin='lower')
```

We are considering 3 directional derivatives:
- $\vec{u}_1 = \hat{i}$
- $\vec{u}_2 = -\hat{j}$
- $\vec{u}_3 = \frac{1}{\sqrt{2}}\hat{i}+\frac{1}{\sqrt{2}}\hat{j} $

```python
grad = np.gradient(T(X,Y))

D1 = grad[1]

D2 = -grad[0]

D3 = 1/np.sqrt(2)*(grad[0] + grad[1])

fig = plt.figure()
ax = fig.add_subplot(111)
ax.imshow(D1, origin='lower')
ax.set_title(r"$D_{\vec{u}_1}$")

fig = plt.figure()
ax = fig.add_subplot(111)
ax.imshow(D2, origin='lower')
ax.set_title(r"$D_{\vec{u}_2}$")

fig = plt.figure()
ax = fig.add_subplot(111)
ax.imshow(D3, origin='lower')
ax.set_title(r"$D_{\vec{u}_3}$")
```

## 2D vector field

```python
X, Y = np.meshgrid(np.arange(0, 2 * np.pi, .2), np.arange(0, 2 * np.pi, .2))
U = np.cos(X)
V = np.sin(Y)

fig = plt.figure()
ax = fig.add_subplot(111)
ax.quiver(X, Y, U, V, units='width')
```

Only consider the directional derivatives in the $\vec{u} = \hat{i}$ direction.

```python
D1x = np.gradient(U)[1]
D1y = np.gradient(V)[1]

fig = plt.figure()
ax = fig.add_subplot(111)
ax.quiver(X, Y, D1x, D1y, units='width')
```
